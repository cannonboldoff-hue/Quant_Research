```python
from statsmodels.tsa.stattools import adfuller

# Define the ADF test function
def adf_test(series, name):
    resuSUNPHARMA = adfuller(series.dropna())
    print(f'ADF Test for {name}:')
    print(f'ADF Statistic: {resuSUNPHARMA[0]}')
    print(f'p-value: {resuSUNPHARMA[1]}')
    print('Critical Values:')
    for key, value in resuSUNPHARMA[4].items():
        print(f'   {key}: {value}')
    if resuSUNPHARMA[1] < 0.05:
        print(f"ResuSUNPHARMA: The time series {name} is stationary.\n")
    else:
        print(f"ResuSUNPHARMA: The time series {name} is non-stationary.\n")

# Perform the ADF test for SUNPHARMA.NS and ONGC.NS
adf_test(df[('Close', 'SUNPHARMA.NS')], 'SUNPHARMA.NS')
adf_test(df[('Close', 'ONGC.NS')], 'ONGC.NS')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Assuming df is already defined with 'SUNPHARMA' and 'ONGC' columns for close prices

# Calculate the ratio of SUNPHARMA to ONGC
ratio = df['SUNPHARMA'] / df['ONGC']

# Calculate the 20-day moving average
moving_average_20 = ratio.rolling(window=20).mean()

# Calculate the Z-score using the ratio and the moving average
z_score = (ratio - moving_average_20) / moving_average_20.std()

# Set risk level
risk_per_trade = 0.5

# Initialize capital and logs for backtesting
initial_capital = 100000  # starting capital
capital = initial_capital
position = 0  # 1 for long, -1 for short, 0 for no position
trade_log = []  # stores each trade's details

# Backtesting logic
for i in range(1, len(df)):
    date = df.index[i]
    price_sunpharma = df['SUNPHARMA'][i]
    price_ONGC = df['ONGC'][i]
    
    # Buying signal: Z-score < -0.5
    if z_score[i] < -0.25 and position == 0:
        position = 1
        entry_price = price_sunpharma / price_ONGC
        trade_log.append((date, 'Buy', entry_price, capital))
    
    # Selling signal: Z-score > 0.5
    elif z_score[i] > 0.25 and position == 1:
        position = 0
        exit_price = price_sunpharma / price_ONGC
        profit = (exit_price - entry_price) * capital * risk_per_trade
        capital += profit
        trade_log.append((date, 'Sell', exit_price, capital))
        
# Convert trade log to DataFrame
trade_log_df = pd.DataFrame(trade_log, columns=['Date', 'Action', 'Price', 'Capital'])

# Calculate cumulative returns
trade_log_df['Cumulative Return'] = trade_log_df['Capital'] - initial_capital

# Calculate Sharpe ratio
daily_returns = trade_log_df['Capital'].pct_change().dropna()
sharpe_ratio = (daily_returns.mean() / daily_returns.std()) * np.sqrt(252)

# Calculate maximum drawdown
trade_log_df['Capital High'] = trade_log_df['Capital'].cummax()
trade_log_df['Drawdown'] = trade_log_df['Capital'] - trade_log_df['Capital High']
trade_log_df['Drawdown %'] = (trade_log_df['Drawdown'] / trade_log_df['Capital High']) * 100
max_drawdown = trade_log_df['Drawdown'].min()
max_drawdown_percentage = trade_log_df['Drawdown %'].min()

# Display metrics
print(f"Total Profit/Loss: {trade_log_df['Cumulative Return'].iloc[-1]:.2f}")
print(f"Sharpe Ratio: {sharpe_ratio:.2f}")
print(f"Max Drawdown: {max_drawdown:.2f}")
print(f"Max Drawdown Percentage: {max_drawdown_percentage:.2f}%")

# Plotting the Z-score and buy/sell signals
fig, ax = plt.subplots(2, 1, figsize=(14, 12))

# Plot Z-score with buy/sell signals
ax[0].plot(z_score, label='Z-Score of SUNPHARMA / ONGC', color='purple')
ax[0].axhline(0.25, color='green', linestyle='--', label='Z-Score = 0.25')
ax[0].axhline(-0.25, color='red', linestyle='--', label='Z-Score = -0.25')
ax[0].legend(loc='upper left')
ax[0].set_title("Z-Score with Buy/Sell Signals")
ax[0].set_xlabel("Date")
ax[0].set_ylabel("Z-Score")

# Mark buy and sell points
for i in range(len(trade_log_df)):
    action = trade_log_df['Action'][i]
    date = trade_log_df['Date'][i]
    price = trade_log_df['Price'][i]
    if action == 'Buy':
        ax[0].plot(date, z_score.loc[date], 'g^', markersize=10)
    elif action == 'Sell':
        ax[0].plot(date, z_score.loc[date], 'rv', markersize=10)

# Plot capital over time with drawdown
ax[1].plot(trade_log_df['Date'], trade_log_df['Capital'], label='Capital', color='blue')
ax[1].fill_between(trade_log_df['Date'], trade_log_df['Capital'], trade_log_df['Capital High'], color='red', alpha=0.3, label='Drawdown')
ax[1].set_title("Capital Over Time with Drawdown")
ax[1].set_xlabel("Date")
ax[1].set_ylabel("Capital")
ax[1].legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from fpdf import FPDF

# Calculate the ratio of SUNPHARMA to ONGC
ratio = df['SUNPHARMA'] / df['ONGC']

# Calculate the 20-day moving average and Z-score
moving_average_20 = ratio.rolling(window=20).mean()
z_score = (ratio - moving_average_20) / moving_average_20.std()

# Initialize capital and position variables for backtesting
initial_capital = 100000
capital_sunpharma = initial_capital
capital_ongc = initial_capital
position_sunpharma = 0  # 1 for long, -1 for short
position_ongc = 0
trade_log_sunpharma = []
trade_log_ongc = []

# Backtesting logic
for i in range(1, len(df)):
    date = df.index[i]
    price_sunpharma = df['SUNPHARMA'][i]
    price_ongc = df['ONGC'][i]
    
    # SUNPHARMA trading logic
    if z_score[i] < -0.25 and position_sunpharma == 0:
        position_sunpharma = 1
        entry_price_sunpharma = price_sunpharma
        trade_log_sunpharma.append((date, 'Buy', entry_price_sunpharma, capital_sunpharma))
    elif z_score[i] > 0.25 and position_sunpharma == 1:
        position_sunpharma = 0
        exit_price_sunpharma = price_sunpharma
        profit_sunpharma = (exit_price_sunpharma - entry_price_sunpharma) * 100  # 100 shares
        capital_sunpharma += profit_sunpharma
        trade_log_sunpharma.append((date, 'Sell', exit_price_sunpharma, capital_sunpharma))

    # ONGC trading logic
    if z_score[i] > 0.25 and position_ongc == 0:
        position_ongc = 1
        entry_price_ongc = price_ongc
        trade_log_ongc.append((date, 'Buy', entry_price_ongc, capital_ongc))
    elif z_score[i] < -0.25 and position_ongc == 1:
        position_ongc = 0
        exit_price_ongc = price_ongc
        profit_ongc = (exit_price_ongc - entry_price_ongc) * 100  # 100 shares
        capital_ongc += profit_ongc
        trade_log_ongc.append((date, 'Sell', exit_price_ongc, capital_ongc))

# Convert trade logs to DataFrames
trade_log_sunpharma_df = pd.DataFrame(trade_log_sunpharma, columns=['Date', 'Action', 'Price', 'Capital'])
trade_log_ongc_df = pd.DataFrame(trade_log_ongc, columns=['Date', 'Action', 'Price', 'Capital'])

# Define function to calculate metrics
def calculate_metrics(trade_log_df, initial_capital):
    trade_log_df['Capital High'] = trade_log_df['Capital'].cummax()
    trade_log_df['Drawdown'] = trade_log_df['Capital'] - trade_log_df['Capital High']
    trade_log_df['Drawdown %'] = (trade_log_df['Drawdown'] / trade_log_df['Capital High']) * 100
    max_drawdown_percentage = trade_log_df['Drawdown %'].min()
    final_capital = trade_log_df['Capital'].iloc[-1]
    profit_percentage = ((final_capital - initial_capital) / initial_capital) * 100
    returns = trade_log_df['Capital'].pct_change().dropna()
    sharpe_ratio = (returns.mean() / returns.std()) * np.sqrt(252)
    return profit_percentage, max_drawdown_percentage, sharpe_ratio

# Calculate metrics for SUNPHARMA
profit_sunpharma, max_drawdown_sunpharma, sharpe_ratio_sunpharma = calculate_metrics(trade_log_sunpharma_df, initial_capital)
# Calculate metrics for ONGC
profit_ongc, max_drawdown_ongc, sharpe_ratio_ongc = calculate_metrics(trade_log_ongc_df, initial_capital)

# Plot drawdown charts for SUNPHARMA and ONGC and save as images
fig, ax = plt.subplots(2, 1, figsize=(10, 10))
ax[0].plot(trade_log_sunpharma_df['Date'], trade_log_sunpharma_df['Drawdown %'], color='blue', label="SUNPHARMA Drawdown %")
ax[0].set_title("SUNPHARMA Drawdown Over Time")
ax[0].set_xlabel("Date")
ax[0].set_ylabel("Drawdown %")
ax[0].legend()
ax[1].plot(trade_log_ongc_df['Date'], trade_log_ongc_df['Drawdown %'], color='purple', label="ONGC Drawdown %")
ax[1].set_title("ONGC Drawdown Over Time")
ax[1].set_xlabel("Date")
ax[1].set_ylabel("Drawdown %")
ax[1].legend()
fig.tight_layout()
drawdown_chart_path = "drawdown_charts.png"
fig.savefig(drawdown_chart_path)

# Generate PDF report
pdf = FPDF()
pdf.add_page()
pdf.set_font("Arial", "B", 16)
pdf.cell(0, 10, "Backtest Report", ln=True, align="C")
pdf.set_font("Arial", "", 12)

# SUNPHARMA Metrics
pdf.cell(0, 10, "SUNPHARMA Metrics:", ln=True)
pdf.cell(0, 10, f"Profit Percentage: {profit_sunpharma:.2f}%", ln=True)
pdf.cell(0, 10, f"Max Drawdown: {max_drawdown_sunpharma:.2f}%", ln=True)
pdf.cell(0, 10, f"Sharpe Ratio: {sharpe_ratio_sunpharma:.2f}", ln=True)
pdf.ln(5)

# ONGC Metrics
pdf.cell(0, 10, "ONGC Metrics:", ln=True)
pdf.cell(0, 10, f"Profit Percentage: {profit_ongc:.2f}%", ln=True)
pdf.cell(0, 10, f"Max Drawdown: {max_drawdown_ongc:.2f}%", ln=True)
pdf.cell(0, 10, f"Sharpe Ratio: {sharpe_ratio_ongc:.2f}", ln=True)
pdf.ln(5)

# SUNPHARMA Trade Log
pdf.cell(0, 10, "SUNPHARMA Trade Log:", ln=True)
for idx, row in trade_log_sunpharma_df.iterrows():
    pdf.cell(0, 10, f"{row['Date']} - {row['Action']} at {row['Price']:.2f} - Capital: {row['Capital']:.2f}", ln=True)

pdf.add_page()
# ONGC Trade Log
pdf.cell(0, 10, "ONGC Trade Log:", ln=True)
for idx, row in trade_log_ongc_df.iterrows():
    pdf.cell(0, 10, f"{row['Date']} - {row['Action']} at {row['Price']:.2f} - Capital: {row['Capital']:.2f}", ln=True)

# Add drawdown chart to PDF
pdf.add_page()
pdf.cell(0, 10, "Drawdown Charts:", ln=True)
pdf.image(drawdown_chart_path, x=10, y=30, w=190)

# Save PDF
pdf_output_path = "backtest_report.pdf"
pdf.output(pdf_output_path)

print("PDF report generated:", pdf_output_path)
```
