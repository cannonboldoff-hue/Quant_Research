```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/merged_file.csv')
df
```

```python
import pandas as pd

# Ensure the Date column is in datetime format
df["Date"] = pd.to_datetime(df["date"])

# Group by Ticker first, then resample in 5-minute intervals
df_resampled = (
    df.groupby("Ticker")
    .resample("5T", on="Date")
    .agg({
        "open": "first",    # First price in the 5-min period
        "high": "max",      # Highest price in the 5-min period
        "low": "min",       # Lowest price in the 5-min period
        "close": "last",    # Last price in the 5-min period
        "volume": "sum"     # Total volume in the 5-min period
    })
    .dropna()
    .reset_index()
)

# Save the 5-minute dataset
df_resampled.to_csv("/content/drive/MyDrive/5min_data.csv", index=False)

print("5-minute dataset saved as 'your_5min_data.csv'")
```
