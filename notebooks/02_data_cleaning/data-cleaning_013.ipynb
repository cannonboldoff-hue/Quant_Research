```python
import pandas as pd

data = pd.read_parquet('/content/drive/MyDrive/ipynb/Quarter_parquet/Quarter_2024.parquet')
data
```

```python
tickers_to_filter = ['HDFCBANK.NSE', 'RELIANCE.NSE', 'NTPC.NSE', 'TATASTEEL.NSE', 'TATAMOTORS.NSE']

data = data[data['Ticker'].isin(tickers_to_filter)]

data
```

```python
import pandas as pd

# Convert 'DateTime' column to datetime format if not already
data['DateTime'] = pd.to_datetime(data['Date_Time'])

# Extract time from DateTime column
data['Time'] = data['DateTime'].dt.time

# Define start and end time
start_time = pd.to_datetime('09:15:00').time()
end_time = pd.to_datetime('15:30:00').time()

# Apply all filters
filtered_data = data[
    (data['Ticker'].isin(tickers_to_filter)) &  # Filter tickers
    (data['Time'] >= start_time) &  # Filter time range
    (data['Time'] <= end_time)  # Filter time range
]

# Drop the temporary 'Time' column
filtered_data = filtered_data.drop(columns=['Time'])

filtered_data.dropna(subset=['LTP',	'BuyPrice','SellPrice'])

filtered_data
```

```python
filtered_data.to_parquet('/content/drive/MyDrive/data/data5.parquet', index=False)
```
