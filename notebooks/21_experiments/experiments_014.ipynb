```python
import pandas as pd
import os
import glob

# Path to the folder containing the CSV files
folder_path = '/content/drive/MyDrive/LTPKAMADSL/Tradebook_LTP'  # change this

# Get all CSV file paths in the folder
csv_files = glob.glob(os.path.join(folder_path, '*.csv'))

# List to store each DataFrame
df_list = []

# Loop through files and read them
for file in csv_files:
    try:
        df = pd.read_csv(file)
        df_list.append(df)
    except Exception as e:
        print(f"Failed to read {file}: {e}")

# Concatenate all DataFrames
merged_df = pd.concat(df_list, ignore_index=True)

# Save the merged DataFrame to a new CSV file
merged_df.to_csv('/content/drive/MyDrive/LTPKAMADSL/Tradebook_LTP/merged_output.csv', index=False)

print("Merging completed. Output saved as 'merged_output.csv'.")
```
