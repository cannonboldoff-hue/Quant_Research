```python
import pandas as pd

# Manually assign column names
df.columns = ['Local time', 'open', 'high', 'low', 'close', 'volume']

# === Step 2: Clean and parse datetime ===
# Remove ' GMT+0530' string and convert to datetime
df['datetime'] = pd.to_datetime(
    df['Local time'].str.replace(r' GMT\+0530', '', regex=True),
    format='%d.%m.%Y %H:%M:%S.%f'
)

# Localize to Asia/Kolkata timezone (GMT+0530)
df['datetime'] = df['datetime'].dt.tz_localize('Asia/Kolkata')

df['Ticker'] = 'NATGASUSD'

# Convert to UTC
df['Datetime'] = df['datetime'].dt.tz_convert('UTC')

# === Step 3: Drop old columns and reorder ===
df = df.drop(columns=['Local time', 'datetime'])

# Reorder columns: datetime_utc first
df = df[['Ticker', 'Datetime', 'open', 'high', 'low', 'close', 'volume']]

# === Step 4: Save to new CSV ===
df.to_csv("/content/drive/MyDrive/Data/NATGASUSD.csv", index=False)
```
