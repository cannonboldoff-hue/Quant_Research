```python
import pandas as pd

# --- Transform second dataset ---
second_df = second_df.rename(columns={
    "Local time": "Datetime",
    "Open": "open",
    "High": "high",
    "Low": "low",
    "Close": "close",
    "Volume": "volume"
})

# Parse datetime with offset (GMT+0530 → UTC)
second_df["Datetime"] = pd.to_datetime(
    second_df["Datetime"],
    format="%d.%m.%Y %H:%M:%S.%f GMT%z"
).dt.tz_convert("UTC")

# Remove timezone info → tz-naive
second_df["Datetime"] = second_df["Datetime"].dt.tz_localize(None)

# Add ticker column
second_df["Ticker"] = "WTIUSD60"

# Reorder columns
second_df = second_df[["Ticker", "Datetime", "open", "high", "low", "close", "volume"]]

# --- Make main_df consistent too ---
main_df["Datetime"] = pd.to_datetime(main_df["Datetime"])
main_df["Datetime"] = main_df["Datetime"].dt.tz_localize(None)

# --- Append ---
combined_df = pd.concat([main_df, second_df], ignore_index=True)

# Sort safely
combined_df = combined_df.sort_values(["Ticker", "Datetime"]).reset_index(drop=True)
```
