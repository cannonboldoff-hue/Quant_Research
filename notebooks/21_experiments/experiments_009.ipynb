```python
import pandas as pd
import numpy as np

# Load the dataset
file_path = '/content/drive/MyDrive/Task 3 and 4_Loan_Data.csv'  # Update with the actual file path
loan_data = pd.read_csv(file_path)
loan_data
```

```python
# Define the number of buckets (adjust as needed)
num_buckets = 5  # For example, create 5 buckets

# Use quantile-based binning to divide FICO scores into buckets
loan_data['Rating'] = pd.qcut(
    loan_data['fico_score'],
    q=num_buckets,
    labels=range(1, num_buckets + 1)  # Lower ratings signify better credit scores
)

# Save the modified dataset with ratings
output_file_path = '/content/drive/MyDrive/Loan_Data_with_Ratings.csv'  # Update with the desired output path
loan_data.to_csv(output_file_path, index=False)

print(f"Ratings added successfully! Output saved to {output_file_path}")
loan_data
```
