```python
import pandas as pd
import glob
import os

# folder path
folder_path = '/content/drive/MyDrive/Comodoties_Data'  # UPDATE this

# list of CSV files in the folder
csv_files = glob.glob(os.path.join(folder_path, '*.csv'))

# list to store individual dataframes
dfs = []

for file in csv_files:
    # Read CSV without header
    df = pd.read_csv(file, header=None, sep=r'\s+', engine='python')

    # Assign proper columns
    df.columns = ['Datetime', 'Time', 'open', 'high', 'low', 'close', 'volume']

    # Combine date and time
    df['Datetime'] = df['Datetime'] + ' ' + df['Time']
    df.drop('Time', axis=1, inplace=True)

    # Add Ticker based on filename (without extension)
    ticker = os.path.splitext(os.path.basename(file))[0]
    df['Ticker'] = ticker

    # Reorder columns
    df = df[['Ticker', 'Datetime', 'open', 'high', 'low', 'close', 'volume']]

    dfs.append(df)

# Combine all dataframes
merged_df = pd.concat(dfs, ignore_index=True)

# OPTIONAL: sort by Ticker and Datetime
merged_df.sort_values(by=['Ticker', 'Datetime'], inplace=True)

# Show the result
print(merged_df.head())

# Save to a new CSV
merged_df.to_csv('/content/drive/MyDrive/Comodoties_Data/Comodoties_Data_merged.csv', index=False)
```
