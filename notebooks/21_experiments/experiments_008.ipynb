```python
import pandas as pd
import os

def merge_csv_files_in_series(directory_path, output_file):
    # List all csv files in the given directory
    csv_files = [f for f in os.listdir(directory_path) if f.endswith('.csv')]
    
    if not csv_files:
        print("No csv files found in the directory.")
        return
    
    # Sort the files lexicographically (modify sorting if required)
    csv_files.sort()
    
    # Initialize an empty list to store DataFrames
    dataframes = []
    
    # Merge the first 21 csv files (if there are fewer, adjust automatically)
    for i, file in enumerate(csv_files[:21]):
        file_path = os.path.join(directory_path, file)
        
        # Read the csv file
        df = pd.read_csv(file_path)
        
        # Append the DataFrame to the list
        dataframes.append(df)
        print(f'Merged file {i+1}/{min(21, len(csv_files))}: {file}')
    
    # Concatenate all DataFrames
    if dataframes:
        merged_df = pd.concat(dataframes, ignore_index=True)
        
        # Write the merged DataFrame to a new csv file
        merged_df.to_csv(output_file, index=False)
        print(f'Merged csv file saved to {output_file}')
    else:
        print("No data was merged.")

# Example usage
directory_path = 'rrq'  # Path to the AUG directory containing csv files
output_file = 'rrq/rrq_2024.csv'  # Path to save the merged csv file
merge_csv_files_in_series(directory_path, output_file)
```
